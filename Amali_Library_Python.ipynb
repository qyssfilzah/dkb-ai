# Amali pandas

```python
import pandas as pd
data = {
    "Nama" : ["Ali", "Siti", "Kumar", "Mei Ling", "Aiman"],
    "Markah" : [70, 85, 90, 65, 80]
}
```

# Bina Dataframe

```python
df = pd.DataFrame(data)
print(df)
```

```python
print(df.head())
print(df["Nama"])
print(df["Markah"])
```

# Kira purata markah

```python
purata = df["Markah"]. mean()
print("Purata markah:", purata)
```

# Tambah Lajur Status

```python
import numpy as np
df['Status'] = np.where(df["Markah"] >= 50, "Lulus", "Gagal")
print(df)
```

# Tambah lajur gred

```python
def tentukan_gred(markah):
    if markah >= 80:
        return "A"
    elif markah >= 70:
        return "B"
    elif markah >= 65:
        return "C"
    elif markah >= 50:
        return "D"
    else:
        return "F" # Add a default grade for marks below 50

df["Gred"] = df["Markah"].apply(tentukan_gred)
print(df)
```

# Simpan data ke dalam fail CSV

```python
df.to_csv("markah_pelajar.csv", index=False)
print("Fail berjaya disimpan.")
```
